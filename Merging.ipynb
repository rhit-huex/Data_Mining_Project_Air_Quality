```python
import pandas as pd
df_k = pd.read_csv("./data/kaggle_cleaned.csv")
df_g = pd.read_csv("./data/EPA_cleaned.csv")
```

```python
df_k.head()
```

```python
df_g.head()
```

```python
# Keep AQI, Category, Date, cbsa
```

```python
# 1. Prepare df_k (Kaggle)
# Create 'cbsa' to match the other dataset and ensure Date is datetime
df_k['cbsa'] = df_k['city_ascii'] + ", " + df_k['state_id']
df_k['Date'] = pd.to_datetime(df_k['Date'])
# Select only the common columns
df_k_subset = df_k[['cbsa', 'Date', 'AQI', 'Category']].copy()

# 2. Prepare df_g (EPA)
# Rename columns to match df_k
df_g = df_g.rename(columns={'aqi': 'AQI', 'category': 'Category', 'date': 'Date'})
df_g['Date'] = pd.to_datetime(df_g['Date'])
# Select only the common columns
df_g_subset = df_g[['cbsa', 'Date', 'AQI', 'Category']].copy()

# 3. Concatenate (Stack) the datasets
# This combines the rows from both, aligning the columns automatically
unified_df = pd.concat([df_k_subset, df_g_subset], ignore_index=True)

# Optional: Drop exact duplicates if there is any accidental overlap
unified_df = unified_df.drop_duplicates()

# Sort by Location and Date to see the continuous timeline
unified_df = unified_df.sort_values(by=['cbsa', 'Date'])

unified_df.head()
```

```python
unified_df.to_csv('./data/merged_dataset.csv')
```
