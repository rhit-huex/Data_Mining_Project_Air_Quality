```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./data/merged_dataset.csv")
df.head(3)
```

```python
df["Date"] = pd.to_datetime(df["Date"])

# Create time features
df["year"] = df["Date"].dt.year
df["month"] = df["Date"].dt.month
df["dayofyear"] = df["Date"].dt.dayofyear
```

```python
columbus = df[df['cbsa'] == 'Columbus, OH'].copy()
```

```python
yearly_avg = columbus.groupby("year")["AQI"].mean()

plt.figure()
yearly_avg.plot()
plt.title("Average Yearly AQI - Columbus, OH")
plt.xlabel("Year")
plt.ylabel("Average AQI")
plt.show()
```
