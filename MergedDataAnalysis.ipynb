```python
import pandas as pd
df = pd.read_csv("./data/merged_dataset.csv")
df.tail()
```

# Ideas for analysis

Do rolling average for all of america

Pick some towns and do specific analysis on them over the years

Calculate average change on AQI or max difference in aqi and put on heatmap



```python
import matplotlib.pyplot as plt

# Filter the data for Yuma, AZ
yuma_df = df[df['cbsa'] == 'Yuma, AZ'].copy()

# Convert the 'Date' column to datetime objects for proper plotting
yuma_df['Date'] = pd.to_datetime(yuma_df['Date'])

# Sort the data by date
yuma_df = yuma_df.sort_values('Date')

# Create the plot
plt.figure(figsize=(14, 7))
plt.plot(yuma_df['Date'], yuma_df['AQI'], label='Daily AQI', alpha=0.6)

# Optional: Add a rolling average (e.g., 30-day) to see the trend more clearly
yuma_df['AQI_Rolling'] = yuma_df['AQI'].rolling(window=30).mean()
plt.plot(yuma_df['Date'], yuma_df['AQI_Rolling'], color='red', linewidth=2, label='30-Day Moving Average')

plt.title('Air Quality Index (AQI) Over Time - Yuma, AZ')
plt.xlabel('Date')
plt.ylabel('AQI')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
